### 1. 探索

条件を満たすマスを探す

例：エッジで白のマスを探す

まずは関数とfor文の使い方の復習&新しい使い方の勉強

```python
# 返り値のある関数の例

def hello() :
    return "HELLO"

def func() :
    print("A")
    return "XXX"
    print("B")

def id2color(i) :
    return cube()[i][1][1]
```

```python
# for文の使い方
for i in range(6) :
    print(cube()[i][1][1])
```

```python
for i in range(1, 5) :
    print(cube()[i][1][1])
```

```python
for i in range(6) :
    for j in range(3) :
        for k in range(3) :
            print(i, j, k)
```

```python
for i in range(6) :
    for j in range(3) :
        for k in range(3) :
            print(cube()[i][j][k])
```

```python
def for_example() :
    for i in range(6) :
        for j in range(3) :
            for k in range(3) :
                if i == 3 and j == 1 :
                    return [i, j, k]

print(for_example())
```

```python
# 6*3*3個あるタイルから白タイルを探す
def search_white() :
    for i in range(6) :
        for j in range(3) :
            for k in range(3) :
                if cube()[i][j][k] == "W" :
                    print(i, j, k)
```
