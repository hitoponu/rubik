#### for文の入れ子

```python
for j in range(3) :
    for k in range(3) :
        print(j, k)
```

```python
for i in range(3) :
    for j in range(3) :
        print(cube()[0][i][j])
```

```python
for i in range(3) :
    for j in range(3) :
        if cube()[0][i][j] == "R" :
            print("R is found")
```

```python
for i in range(6) :
    for j in range(3) :
        for k in range(3) :
            print(cube()[i][j][k])
```

```python
# 6*3*3個あるタイルから白タイルを探す
def locate_white() :
    for i in range(6) :
        for j in range(3) :
            for k in range(3) :
                if cube()[i][j][k] == "W" :
                    print(i, j, k)
                    
def search_white() :
    for i in range(6) :
        for j in range(3) :
            for k in range(3) :
                if cube()[i][j][k] == "W" :
                    return [i, j, k]
```

1. 別の色を探す関数を作ろう
2. エッジの白マスを探す関数を作ろう
3. 探索範囲を変えてみよう

ここまででできることの例：

* 白の面が揃っているかをfor文を使って判定
* 4つあるエッジの白マスの位置を全てprint
* 正しい位置にない白のマスを1つ探して返す関数

### 相対的な操作

今は緑を前面とした記号の操作しかない。サイドの面(緑、赤、青、橙)を番号で指定することで、相対的な操作が可能になる。

```python
do("R") # R()　と同じ
```

```python
sides = ["F", "R", "B", "L"]
```

```python
do(sides[0])
```

```python
#　xの番号の面の右の列を上面に持ってくる
x = 3
do(sides[(x + 1) % 4])
```

1. 正面の面の番号を引数に、好きな技を相対的に実行する関数を作ろう
2. 2段目の白マスをsearchし、底面へ移動させよう。

ここまででできることの例：

* 白のエッジをsearchし、その位置に応じて相対的な操作をすることで、その白マスを底面へ移動する
