## 複数の操作をまとめて関数にしよう

今私たちがルービックキューブを操作するときは`R()`のような一手動かす関数を使っています。

例えば、セクシームーブのような1つの技を一気に実行する関数があれば楽になりそうです。

関数により複数の操作を1まとめにすることができます。

下のセルを実行してみましょう。

````python
# セクシームーブの関数の例
def sexy_move() :
    R()
    U()
    Ri()
    Ui()
````

上はセクシームーブを行う関数を定義するプログラムです。

`sexy_move`というのは関数名です。

```
    R()
    U()
    Ri()
    Ui()
```
はこの関数が呼び出された時に行われる操作です。

if文やfor文同様インデントが必要なことに気をつけましょう。

関数は以下のように、関数名を指定することで呼び出せます。

下のセルを実行してみましょう。

````python
sexy_move()
````

上の`sexy_move()`の呼び出しでRUR'U'の操作が行われます。

ここまでで、関数の定義と関数の呼び出し方を学びました。

## 引数によって関数の処理を変える

1秒待つ関数sleep(1)は()の中の数字によって待つ時間が変化しました。

()の中に与えられる値を`引数`と言い、引数によって関数の処理を変えることができます。

下のセルを実行してみましょう。

````python
def repeat_sexy_move(n) :
    for i in range(n) : # n回繰り返す
        sexy_move()
````

上はセクシームーブを`引数`の数だけ繰り返す関数です。

引数がある関数を定義するときは、関数名の横の()の中に変数を書きます。

関数の中の処理でこの変数を利用することができます。（この例ではfor文の繰り返し回数に利用されている）

引数のある関数の呼び出しでは以下のように()の中に引数の値を書きます。

````python
repeat_sexy_move(3)
````
