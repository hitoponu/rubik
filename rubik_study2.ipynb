### while文

条件を満たす間同じ操作を繰り返す

```python
x = 2
while x < 1000 :
    print(x)
    x = x * x
```

```python
n = 100
while not n == 1 :
    if n % 2 == 0 :
        n = n // 2
    else :
        n = 3 * n + 1
    print(n)
```
