## Step 7：段階1を完成させよう（合体）

1回で全部の白がデイジーに入るとは限りません。そこで、

* Step 3 の判定で**デイジーが完成していたら終わり**（`break`）
* そうでなければ、場合A〜Eの処理をもう一度

を、完成するまで繰り返します。

> `while True :` を使って、これまでのプログラムを合体させよう。
>
> 形のヒント：
> ```python
> while True :
>     # Step 3：デイジーが完成したか調べる。完成していたら break
>
>     # Step 5：場合A
>     # Step 6：場合B
>     # Step 6：場合C
>     # Step 6：場合D
>     # Step 6：場合E
> ```
>
> 何回か `shuffle()` して、毎回デイジーができるか確かめよう。
