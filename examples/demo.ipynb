セルを1つずつ実行していくと、3Dの窓の中身が変わっていく。
窓は**別のウィンドウ**として開く (ノートの中には出ない)。

## いちばん手軽な使いかた

`rubik.R()` のように引数を省くと、rubik が持っている1つのキューブが回る。
最初の1手のところで3Dの窓がひとりでに開く。

```python
import rubik

rubik.R()
```

**窓が開きましたか。** マウスでドラッグして、好きな向きにしてから次に進んでください。

## 1手ずつ回す

このセルは何度実行してもよい。実行するたびに窓の中身が変わるが、
**さっきマウスで決めた向きは変わらない**ことを確かめてほしい。

```python
rubik.U()
```

## 手順をまとめて

`do()` には、本や Web に載っているままの書きかたを渡せる。
空白は無くてよい。`R U R' U'` は6回くり返すと元に戻る。

```python
rubik.solved()                  # いったん完成状態に
rubik.do("RUR'U'", times=6)

rubik.is_solved()
```

## いまの状態を見る

`rubik.cube()` で、いまのキューブの 6x3x3 リストが取れる。

回す・そろえる・混ぜるといった**状態を変える呼び出しは、何も返さない**。
セルに書いてもリストがだらだら表示されないようにするため。
中身を見たいときは `rubik.cube()` か `rubik.show()` を使う。

```python
rubik.shuffle(seed=1)    # 何も表示されない
rubik.show()             # 展開図で見る
```

```python
rubik.cube()[2]          # F面 (前) の 3x3
```

`rubik.cube(x)` と書けば、丸ごと差しかえられる。
このときだけは窓が追随しないので、`rubik.update()` で描き直す。

```python
rubik.cube(rubik.Cube().cube())   # 完成状態に差しかえる
rubik.update()                      # 窓に反映する
```

## 向きを決まった位置に戻す

```python
rubik.reset_DBL()   # 下・後・左 が見える向き
```

```python
rubik.reset_UFR()   # 上・前・右 が見える向き (最初の向き)
```

## キューブを自分で作る

`rubik.Cube()` で、独立したキューブをいくつでも作れる。
`show3d=True` にすると、そのキューブ専用の窓が開く。

セルの最後に `a` とだけ書くと、展開図が表示される。

```python
a = rubik.Cube(show3d=True)
a.shuffle(seed=1)
a
```

```python
b = rubik.Cube(show3d=True)   # 2つめの窓
b.do("RUR'U'")
b
```

## 中段を回す

外側の面だけでなく、**まん中の層**も回せる。

| 名前 | 回る層 | 向き |
|---|---|---|
| `M` | L と R のあいだ（縦） | `L` と同じ |
| `E` | U と D のあいだ（横） | `D` と同じ |
| `S` | F と B のあいだ（奥ゆき） | `F` と同じ |

中段には角の小立方体が無いので、**回しても角は動かない**。
かわりに中心のステッカーが動くので、面の色そのものがずれる。

```python
rubik.solved()
rubik.M()        # まん中の縦の層を回す
rubik.show()
```

```python
rubik.solved()
rubik.do("M2E'S")   # 手順の中でも使える
rubik.show()
```

## 窓を閉じる

**`rubik.wait()` は呼ばないこと。** ノートブックでは、カーネルが
窓を閉じるまで止まってしまう。閉じたいときは `close()` を使う。

```python
rubik.close()
a.close()
b.close()
```
