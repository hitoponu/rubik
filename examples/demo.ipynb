# rubik を Jupyter Notebook で使う

セルを1つずつ実行していくと、3Dの窓の中身が変わっていく。
窓は**別のウィンドウ**として開く (ノートの中には出ない)。
ブラウザのうしろに隠れていたら、前に出して並べておくとよい。

## いちばん手軽な使いかた

`rubik.R()` のように引数を省くと、rubik が持っている1つのキューブが回る。
最初の1手のところで3Dの窓がひとりでに開く。

```python
import rubik

rubik.R()
```

**窓が開きましたか。** マウスでドラッグして、好きな向きにしてから次に進んでください。

## 1手ずつ回す

このセルは何度実行してもよい。実行するたびに窓の中身が変わるが、
**さっきマウスで決めた向きは変わらない**ことを確かめてほしい。

```python
rubik.U()
```

## 手順をまとめて

`do()` には、本や Web に載っているままの書きかたを渡せる。
空白は無くてよい。`R U R' U'` は6回くり返すと元に戻る。

```python
rubik.solved()                  # いったん完成状態に
rubik.do("RUR'U'", times=6)
rubik.show()
```

## いまの状態を見る

`rubik.cube` が、いまのキューブの 6x3x3 リスト。そのまま読み書きできる。

```python
rubik.cube[2]        # F面 (前) の 3x3
```

```python
rubik.show()         # 展開図
```

`rubik.cube` を直接書きかえたときは、窓は追随しない。
描き直したければ `rubik.update()` を呼ぶ。

```python
rubik.cube = rubik.shuffle(seed=1)
rubik.update()
```

## 向きを決まった位置に戻す

```python
rubik.reset_DBL()   # 下・後・左 が見える向き
```

```python
rubik.reset_UFR()   # 上・前・右 が見える向き (最初の向き)
```

## キューブを自分で作る

`rubik.Cube()` で、独立したキューブをいくつでも作れる。
`show3d=True` にすると、そのキューブ専用の窓が開く。

```python
a = rubik.Cube(show3d=True)
a.shuffle(seed=1)
a
```

```python
b = rubik.Cube(show3d=True)   # 2つめの窓
b.do("RUR'U'")
b
```

## 窓を閉じる

**`rubik.wait()` は呼ばないこと。** ノートブックでは、カーネルが
窓を閉じるまで止まってしまう。閉じたいときは `close()` を使う。

```python
rubik.close()
a.close()
b.close()
```
