# rubik を Jupyter Notebook で使う

セルを1つずつ実行していくと、3Dの窓の中身が変わっていく。
窓は**別のウィンドウ**として開く (ノートの中には出ない)。
ブラウザのうしろに隠れていたら、前に出して並べておくとよい。

## キューブを作る

まずは完成状態を作って、展開図を文字で見てみる。

```python
import rubik

cube = rubik.solved()
rubik.show(cube)
```

## 3Dの窓を開く

別ウィンドウが開く。マウスでドラッグすると、ぐるぐる回せる。
**好きな向きにしてから、次のセルに進んでみてください。**

```python
rubik.init(cube)
```

## 1手ずつ回す

このセルは何度実行してもよい。実行するたびに窓の中身が変わるが、
**さっきマウスで決めた向きは変わらない**ことを確かめてほしい。

```python
cube = rubik.R(cube)
rubik.update(cube)

rubik.show(cube)
```

## ぐちゃぐちゃにする

`seed` に数を渡すと、何度やっても同じ配置になる。

```python
cube = rubik.shuffle(seed=1)
rubik.update(cube)
```

## 向きを決まった位置に戻す

回しすぎて分からなくなったときに使う。

```python
rubik.reset_DBL()   # 下・後・左 が見える向き
```

```python
rubik.reset_UFR()   # 上・前・右 が見える向き (最初の向き)
```

## 窓を閉じる

**`rubik.wait()` は呼ばないこと。** ノートブックではカーネルが
窓を閉じるまで止まってしまう。閉じたいときは `close()` を使う。

```python
rubik.close()
```
